```python
import os

image_dir = r"C:\Users\claud\Desktop\Proiect_Varianta 1\images\train"
label_dir = r"C:\Users\claud\Desktop\Proiect_Varianta 1\labels\train"

def rename_everything_cleanly(img_path, lbl_path):
    
    images = [f for f in os.listdir(img_path) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
    print(f"Am găsit în total {len(images)} imagini.")

    count = 1
    for old_img_name in images:
        
        base_name = os.path.splitext(old_img_name)[0]
        img_ext = os.path.splitext(old_img_name)[1]
        
        
        old_lbl_name = base_name + ".txt"
        
        old_img_full = os.path.join(img_path, old_img_name)
        old_lbl_full = os.path.join(lbl_path, old_lbl_name)
        
        
        if os.path.exists(old_lbl_full):
            
            new_name = str(count)
            new_img_full = os.path.join(img_path, new_name + img_ext)
            new_lbl_full = os.path.join(lbl_path, new_name + ".txt")
            
            
            os.rename(old_img_full, new_img_full)
            os.rename(old_lbl_full, new_lbl_full)
            
            print(f"Succes: [{old_img_name}] s-a transformat în [{new_name}{img_ext}]")
            count += 1
        else:
            print(f"Sărit (lipsă label): {old_img_name}")

    print(f"\n✅ Gata! Acum ai {count-1} perechi ordonate numeric.")


rename_everything_cleanly(image_dir, label_dir)
```

```python
import os
import shutil


base_path = r"C:\Users\claud\Desktop\Proiect_Varianta 1"
original_img = os.path.join(base_path, "images", "train")
original_lbl = os.path.join(base_path, "labels", "train")

new_img_folder = os.path.join(base_path, "images", "train_CURAT")
new_lbl_folder = os.path.join(base_path, "labels", "train_CURAT")

os.makedirs(new_img_folder, exist_ok=True)
os.makedirs(new_lbl_folder, exist_ok=True)


mapping = {'5': '0', '4': '1'}

def curatare_totala():
   
    lista_imagini = [f for f in os.listdir(original_img) if f.endswith('.png')]
    lista_imagini.sort()

    count = 1
    for img_name in lista_imagini:
        nume_baza = os.path.splitext(img_name)[0]
        cale_img_veche = os.path.join(original_img, img_name)
        cale_lbl_veche = os.path.join(original_lbl, nume_baza + ".txt")

        
        if os.path.exists(cale_lbl_veche):
            
            with open(cale_lbl_veche, 'r') as f:
                lines = f.readlines()
            
            new_lines = []
            for line in lines:
                parts = line.split()
                if len(parts) > 0 and parts[0] in mapping:
                    parts[0] = mapping[parts[0]] 
                    new_lines.append(" ".join(parts) + "\n")
                else:
                    new_lines.append(line)

            
            with open(os.path.join(new_lbl_folder, f"{count}.txt"), 'w') as f:
                f.writelines(new_lines)
            
            
            shutil.copy2(cale_img_veche, os.path.join(new_img_folder, f"{count}.png"))
            
            print(f"Sincronizat: {img_name} -> {count}.png")
            count += 1

    print(f"\n✅ GATA! Am creat {count-1} perechi perfecte în folderele '_CURAT'.")

curatare_totala()
```
